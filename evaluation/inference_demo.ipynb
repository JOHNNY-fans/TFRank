```python
# 假设你已经在本地启动了 vLLM 服务，例如：
# python -m vllm.entrypoints.openai.api_server \
#   --model /path/to/your/checkpoint \
#   --port 8113 \
#   --served-model-name rele_pointwise
# such as using vllm_server_init_8113.sh

from tfrank_demo import TFRankDemoRanker
```

```python
# 填写你自己的模型路径
model_path = "Johnnyfans/TFRank-GRPO-Qwen3-0.6B"

ranker = TFRankDemoRanker(
    model_name=model_path,
    api_base="http://localhost:8113/v1",  # 与 vLLM 启动参数对应
    api_key="any-string",                 # vLLM 默认不会校验
    think_mode=False,                     # 默认用 no-think 模式
    reasoning_model=True,                # 如果是 reasoning 版模型就改成 True，训练后的 TFRank 均为 reasoning 版模型
    max_new_tokens=256,
    temperature=0.0,
)
```

```python
query = "what nano means"
doc = "What does nano mean? Nano means very, very small. When it comes to making your body work, nano-materials are very important. A nanometre is one millionth of a millimetre. Your fingernail is about one millimetre thick. There are a lot of nano-materials making up your finger nail! Nanotechnology scientists move atoms and molecules around to make amazing new technologies. Nanotechnology is already in products like sunscreen."
print("Query:", query)
print("Doc:", doc)
```

```python
final_score, fg_score, yes_score, prompt, response = ranker.score(query, doc)

print(f"Final relevance score      (0-1): {final_score:.4f}")
print(f"Fine-grained score (norm.) (0-1): {fg_score:.4f}")
print(f"Yes-probability            (0-1): {yes_score:.4f}")
print("Raw input prompt:", prompt)
print("Raw model response:", response)
```
